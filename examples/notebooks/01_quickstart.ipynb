## Setup

First, ensure the package is installed:

## Example 1: GridWorld Navigation

```python
# Create a gridworld task
spec = TaskSpec(
    environment="gridworld",
    initial_state={"x": 0, "y": 0, "goal": {"x": 5, "y": 5}},
    reward_spec={"id": "reward:v1"},
    termination={"max_steps": 50},
    seed=12345
)

# Execute
uti = UniversalTaskInterface(seed=12345)
result = uti.execute_task(spec, use_adapter=True)

print(f"Steps taken: {len(result.action_sequence)}")
print(f"Total reward: {result.cumulative_reward:.2f}")
print(f"Accuracy: {result.metrics['accuracy']:.2f}")
print(f"Complexity: {result.metrics['complexity_level']}")
```

## Example 2: Task Complexity Estimation

```python
# Estimate complexity of different tasks
simple_task = TaskSpec(
    environment="simple",
    initial_state={"x": 0},
    reward_spec={"id": "reward:v1"},
    termination={"max_steps": 10},
)

complex_task = TaskSpec(
    environment="complex",
    initial_state={f"var_{i}": 0 for i in range(50)},
    reward_spec={"id": "reward:v1", "params": {f"p_{i}": i for i in range(10)}},
    termination={"max_steps": 1000},
)

simple_score, simple_level = estimate_task_complexity(simple_task)
complex_score, complex_level = estimate_task_complexity(complex_task)

print(f"Simple task: {simple_score:.1f} ({simple_level.value})")
print(f"Complex task: {complex_score:.1f} ({complex_level.value})")
```

## Example 3: Meta-Policy Router

```python
# Initialize meta-policy router
router = MetaPolicyRouter(seed=12345)

# Check initial probability distribution
print("Initial strategy probabilities:")
probs = router.get_probability_distribution()
for strategy, prob in probs.items():
    print(f"  {strategy}: {prob:.3f}")

# Measure (collapse superposition)
selected = router.measure()
print(f"\nSelected strategy: {selected}")

# Get hyperparameters for selected strategy
hyperparams = router.get_hyperparams(selected)
print(f"Hyperparameters: {hyperparams}")
```

## Example 4: MAML Adaptation

```python
# Adapt with MAML
task_data = [(i, i**2) for i in range(10)]
adapted_params = router.adapt_with_maml("task1", task_data)

print("Adapted parameters:")
for key, value in adapted_params.items():
    print(f"  {key}: {value:.4f}")

# Update performance
router.update_strategy_performance("maml", 0.85, success=True)

# Get performance stats
stats = router.get_performance_stats()
print(f"\nMAML performance: {stats['maml']}")
```

## Example 5: Multi-Armed Bandit

```python
# Create bandit task
bandit_spec = TaskSpec(
    environment="bandit",
    initial_state={"arm_means": [0.5, 0.3, 0.7, 0.4], "pulls": 0},
    reward_spec={"id": "reward:v1"},
    termination={"max_steps": 20},
    seed=42
)

result = uti.execute_task(bandit_spec, use_adapter=True)

print(f"Total reward: {result.cumulative_reward:.2f}")
print(f"Average reward per pull: {result.cumulative_reward / len(result.action_sequence):.2f}")
```
